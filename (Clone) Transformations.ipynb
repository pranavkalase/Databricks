```python
# Define schema for nested json
from pyspark.sql.types import *
data = [
 (1,  {"f_name": "Dinesh", "l_name": "NY"},10),
(2,  {"f_name": "Rohit", "l_name": "LA"},20),
(3,  {"f_name": "Harsahd", "l_name": "SF"},30),
(4, {"f_name": "Sonali", "l_name": "CHI"},40),
(5, {"f_name": "pari", "l_name": "SEA"},50),
(6,  {"f_name": "Pranav", "l_name": "BOS"},60),
]

schema = StructType([
 StructField('id', IntegerType()),
 StructField('name', StructType([
  StructField('f_name', StringType()),
  StructField('l_name', StringType())
 ])),
 StructField('age', IntegerType())
])
df = spark.createDataFrame(data=data,schema=schema)
display(df)
from pyspark.sql.functions import *
df.select(col('id'),col('name.f_name'),col('name.l_name'),col('age')).display()
```
